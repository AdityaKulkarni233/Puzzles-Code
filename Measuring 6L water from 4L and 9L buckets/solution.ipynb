```python
from collections import deque

capacity_9 = 9
capacity_4 = 4
target = 6

queue = deque([((0, 0), [])])
visited = set()
```

```python
def get_next_states(bucket, jug):
    next_states = []

    # Fill 9L bucket
    next_states.append(((capacity_9, jug), "Fill 9L bucket"))

    # Fill 4L jug
    next_states.append(((bucket, capacity_4), "Fill 4L jug"))

    # Empty 9L bucket
    next_states.append(((0, jug), "Empty 9L bucket"))

    # Empty 4L jug
    next_states.append(((bucket, 0), "Empty 4L jug"))

    # Pour 9L -> 4L
    amount = min(bucket, capacity_4 - jug)
    next_states.append(((bucket - amount, jug + amount), "Pour 9L -> 4L"))

    # Pour 4L -> 9L
    amount = min(jug, capacity_9 - bucket)
    next_states.append(((bucket + amount, jug - amount), "Pour 4L -> 9L"))

    return next_states
```

```python
solution = None

while queue:
    (bucket, jug), steps = queue.popleft()

    if (bucket, jug) in visited:
        continue

    visited.add((bucket, jug))

    # Goal check
    if bucket == target:
        solution = (steps, (bucket, jug))
        break

    for state, action in get_next_states(bucket, jug):
        if state not in visited:
            queue.append((state, steps + [action]))
```

```python
if solution:
    steps, final_state = solution
    print("Solution found:\n")
    for i, step in enumerate(steps, 1):
        print(f"{i}. {step}")
    print("\nFinal state:", final_state)
else:
    print("No solution found.")
```

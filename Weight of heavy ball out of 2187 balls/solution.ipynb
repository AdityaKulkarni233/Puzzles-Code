#### Heavy Ball Puzzle
#### 2187 balls, exactly one is heavier

```python
# Heavy Ball Puzzle
# 2187 balls, exactly one is heavier

balls = list(range(1, 2188))
weighing_count = 0

while len(balls) > 1:

    weighing_count += 1

    size = len(balls) // 3

    group1 = balls[:size]
    group2 = balls[size:2 * size]
    group3 = balls[2 * size:]

    print("Weighing:", weighing_count)
    print("Group 1:", len(group1), "balls")
    print("Group 2:", len(group2), "balls")

    result = input("Enter result (left/right/equal): ").lower()

    if result == "left":
        balls = group1

    elif result == "right":
        balls = group2

    elif result == "equal":
        balls = group3

    else:
        print("Invalid input. Enter left, right, or equal.")
        weighing_count -= 1
        continue

    print("Balls remaining:", len(balls))
    print()

print("Heavy ball:", balls[0])
print("Total weighings:", weighing_count)
```

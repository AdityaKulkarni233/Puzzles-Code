# Poisonous Plants (HackerRank)

## Problem Statement

There are a number of plants in a garden, represented by an array `p` where `p[i]` indicates the pesticide level of the $i^{\text{th}}$ plant. Every day, any plant that has **strictly more pesticide than the plant immediately to its left** dies. 

Determine the total number of days after which no more plants die (i.e., when no plant has a higher pesticide level than the plant to its left).

---

### Examples

#### Example 1
* **Input:** `p = [6, 5, 8, 4, 7, 10, 9]`
* **Output:** `2`
* **Explanation:**
  * **Day 0:** `[6, 5, 8, 4, 7, 10, 9]`
  * **Day 1:** Plants at indices 2 (8 > 5), 4 (7 > 4), and 5 (10 > 7) die. Remaining: `[6, 5, 4, 9]`.
  * **Day 2:** Plant at index 3 (9 > 4) dies. Remaining: `[6, 5, 4]`.
  * **Day 3:** No plant has more pesticide than its left neighbor (`6 > 5 > 4`). Plants stop dying after **2** days.

---

### Constraints
* $1 \le n \le 10^5$
* $0 \le p[i] \le 10^9$

---

## Optimal Solution (Monotonic Stack)

### Intuition & Approach
Simulating day-by-day directly can take $\mathcal{O}(n^2)$ time in the worst case (e.g., strictly increasing arrays), causing Time Limit Exceeded (TLE).

Instead, we can use a **Monotonic Stack** to track the number of days each plant survives before it dies:
1. Maintain a stack of pairs `(pesticide_level, days_to_die)`.
2. For each plant `p[i]`:
   - It will die due to a plant to its left that has a **less than or equal** pesticide level.
   - While the stack is not empty and the top plant has a pesticide level **greater than or equal** to `p[i]`, `p[i]` will outlive that plant. We pop it and track the maximum days `p[i]` needs to wait: `days = max(days, stack_top.days)`.
   - If the stack is empty, it means no plant to the left is smaller than `p[i]`. Thus, `p[i]` will **never die** (`days = 0`).
   - Otherwise, `p[i]` will die 1 day after all smaller/equal plants between it and the smaller plant are eliminated (`days = days + 1`).
3. The overall answer is the maximum `days` recorded across all plants.

```python
def poisonousPlants(p: list[int]) -> int:
    stack = []  # Stack stores tuples: (pesticide_level, days_to_die)
    max_days = 0
    
    for x in p:
        days = 0
        while stack and stack[-1][0] >= x:
            days = max(days, stack.pop()[1])
            
        if not stack:
            days = 0
        else:
            days += 1
            
        max_days = max(max_days, days)
        stack.append((x, days))
        
    return max_days
```

```python
sol = poisonousPlants([6, 5, 8, 4, 7, 10, 9])
print(sol)
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Every plant is pushed and popped from the stack at most once.
* **Space Complexity:** $\mathcal{O}(n)$ — For storing plant records in the monotonic stack.

### All possible combinations 

```python
from itertools import combinations

people = {
    "A": 1,
    "B": 2,
    "C": 7,
    "D": 10
}

times = []


def cross(left, right, torch, total):
    if len(left) == 0:
        times.append(total)
        return

    if torch == "left":
        for pair in combinations(left, 2):

            crossing_time = max(people[pair[0]], people[pair[1]])

            new_left = [p for p in left if p not in pair]
            new_right = right + list(pair)

            cross(
                new_left,
                new_right,
                "right",
                total + crossing_time
            )

    else:
        for person in right:

            new_right = [p for p in right if p != person]
            new_left = left + [person]

            cross(
                new_left,
                new_right,
                "left",
                total + people[person]
            )


cross(["A", "B", "C", "D"], [], "left", 0)

times = sorted(set(times))

print("All possible times:")
print(times)

print("\nMinimum time:", min(times), "minutes")
print("Maximum time:", max(times), "minutes")
```

# 1. Two Sum

## Problem Statement

Given an array of integers `nums` and an integer `target`, return *indices of the two numbers such that they add up to `target`*.

You may assume that each input would have **exactly one solution**, and you may not use the same element twice.

You can return the answer in **any order**.

---

### Examples

#### Example 1
* **Input:** `nums = [2,7,11,15]`, `target = 9`
* **Output:** `[0,1]`
* **Explanation:** Because `nums[0] + nums[1] == 9`, we return `[0, 1]`.

#### Example 2
* **Input:** `nums = [3,2,4]`, `target = 6`
* **Output:** `[1,2]`

#### Example 3
* **Input:** `nums = [3,3]`, `target = 6`
* **Output:** `[0,1]`

---

### Constraints
* $2 \le \text{nums.length} \le 10^4$
* $-10^9 \le \text{nums}[i] \le 10^9$
* $-10^9 \le \text{target} \le 10^9$
* **Only one valid answer exists.**

---

## Optimal Solution (One-Pass Hash Map)

### Intuition & Approach
For every number $x$ in `nums`, we need to find if its complement $\text{complement} = \text{target} - x$ has already been seen in the array.

1. **Hash Map Storage:**  
   Maintain a hash map (or dictionary) that maps a number to its index: `{ value: index }`.

2. **One-Pass Lookups:**  
   As we iterate through `nums` with index `i` and value `num`:
   - Compute `complement = target - num`.
   - If `complement` exists in our hash map, we have found the matching pair! Return `[map[complement], i]`.
   - If not, store the current number and its index in the map (`map[num] = i

```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        n = len(nums)
        res = []
        for i in range(n-1):
            rem = target - nums[i]
            for j in range(i+1,n):
                if rem == nums[j]:
                    res.append(i)
                    res.append(j)
                    return res
        return res
```

```python
sol=Solution()
res= sol.twoSum([2,7,11,15],9)
print(res)
```

```python
sol=Solution()
res= sol.twoSum([3,2,4],6)
print(res)
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — We traverse the list containing $n$ elements only once. Lookups in the hash map take $\mathcal{O}(1)$ average time.
* **Space Complexity:** $\mathcal{O}(n)$ — The extra space required depends on the number of items stored in the hash map, which stores at most $n$ elements.

## Brute Force Approach

### Intuition & Strategy
The simplest way to solve **1. Two Sum** is to test every possible pair of elements in the array and check if their sum equals the target.

1. Use two nested loops:
   - Outer loop picks element `nums[i]` at index `i` from `0` to `n - 2`.
   - Inner loop picks element `nums[j]` at index `j` from `i + 1` to `n - 1`.
2. Check if `nums[i] + nums[j] == target`.
3. As soon as a matching pair is found, return `[i, j]`.

---


```python
class Solution:
    def twoSum(self, nums: list[int], target: int) -> list[int]:
        n = len(nums)
        for i in range(n):
            for j in range(i + 1, n):
                if nums[i] + nums[j] == target:
                    return [i, j]
        return []
```

```python
sol=Solution()
res= sol.twoSum([2,7,11,15],9)
print(res)
```

```python
sol=Solution()
res= sol.twoSum([3,2,4],6)
print(res)
```

```python
sol=Solution()
res= sol.twoSum([3,3],6)
print(res)
```


### Complexity Analysis

* **Time Complexity:** $\mathcal{O}(n^2)$  
  For an array of size $n$, the outer loop runs $n$ times and the inner loop runs on average $n/2$ times. The total number of comparisons is:
  $$\frac{n(n - 1)}{2} \in \mathcal{O}(n^2)$$

* **Space Complexity:** $\mathcal{O}(1)$  
  No extra data structures are used. The algorithm operates in constant auxiliary space.

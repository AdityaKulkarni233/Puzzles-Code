# Mars Exploration (HackerRank)

## Problem Statement

A space explorer's ship crashed on Mars and sends a series of SOS messages back to Earth. Each original SOS message consists of 3 characters: `"SOS"`. 

During transmission through space, cosmic radiation alters some of the characters. Given the received string `s`, calculate total number of letters altered during transmission.

---

### Examples



#### Example 1
* **Input:** `s = "SOSSPSSQSSOR"`
* **Output:** `3`
* **Explanation:**  
  * Expected signal: `SOSSOSSOSSOS`
  * Received signal: `SOSSPSSQSSOR`
  * Differences at index 4 (`P` instead of `O`), index 7 (`Q` instead of `S`), and index 11 (`R` instead of `S`). Total = 3.

#### Example 2
* **Input:** `s = "SOSSOT"`
* **Output:** `1`
* **Explanation:** `SOSSOT` has 1 difference at index 5 (`T` instead of `S`).

#### Example 3
* **Input:** `s = "SOSSOSSOS"`
* **Output:** `0`
* **Explanation:** No characters were altered.

---

### Constraints
* $1 \le \vert{}s\vert{} \le 99$
* $\vert{}s\vert{} \pmod 3 = 0$
* `s` contains only uppercase English letters (`A`-`Z`).

---

## Optimal Solution (Modulo / Index Match)

### Intuition & Approach
Since every sequence of 3 characters is expected to be `"SOS"`:
* Index $i \pmod 3 = 0$ should be `'S'`.
* Index $i \pmod 3 = 1$ should be `'O'`.
* Index $i \pmod 3 = 2$ should be `'S'`.

```python
def marsExploration(s: str) -> int:
    target = "SOS"
    altered_count = 0
    
    for i in range(len(s)):
        if s[i] != target[i % 3]:
            altered_count += 1
            
    return altered_count
```

```python
sol = marsExploration("SOSSPSSQSSOR")
print(sol)
```

```python
sol = marsExploration("SOSSOSSOS")
print(sol)  
```

We can loop through string `s` using character index $i$ and increment our mutated character count whenever `s[i]` does not match the expected character for index position $i \pmod 3$.

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Single pass scan across the string of length $n = \vert{}s\vert{}$.
* **Space Complexity:** $\mathcal{O}(1)$ — Constant space used.

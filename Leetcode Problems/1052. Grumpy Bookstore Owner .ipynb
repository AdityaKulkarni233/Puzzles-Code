# 1052. Grumpy Bookstore Owner

## Problem Statement

There is a bookstore owner that has a store open for $n$ minutes. You are given an integer array `customers` of length $n$ where `customers[i]` is the number of customers that enter the store at the start of the $i^{\text{th}}$ minute and all those customers leave after the end of that minute.

During certain minutes, the bookstore owner is grumpy. You are given a binary array `grumpy` where `grumpy[i]` is `1` if the bookstore owner is grumpy during the $i^{\text{th}}$ minute, and is `0` otherwise.

When the bookstore owner is grumpy, the customers entering during that minute are not satisfied. Otherwise, they are satisfied.

The bookstore owner knows a secret technique to remain **not grumpy** for `minutes` consecutive minutes, but this technique can only be used **once**.

Return the *maximum number of customers* that can be satisfied throughout the day.


### Examples

#### Example 1
* **Input:** `customers = [1,0,1,2,1,1,7,5]`, `grumpy = [0,1,0,1,0,1,0,1]`, `minutes = 3`
* **Output:** `16`
* **Explanation:**  
  The bookstore owner keeps themselves not grumpy for the last 3 minutes (`minutes 5, 6, 7`).  
  The maximum number of satisfied customers = $1 + 1 + 1 + 1 + 7 + 5 = 16$.

#### Example 2
* **Input:** `customers = [1]`, `grumpy = [0]`, `minutes = 1`
* **Output:** `1`



### Constraints
* $n == \text{customers.length} == \text{grumpy.length}$
* $1 \le \text{minutes} \le n \le 2 \times 10^4$
* $0 \le \text{customers}[i] \le 1000$
* `grumpy[i]` is either `0` or `1`.



## Optimal Solution (Sliding Window / Greedy)

### Intuition & Approach

1. **Baseline Satisfied Customers:**  
   Calculate the sum of all customers who are naturally satisfied when `grumpy[i] == 0`. These customers will always be satisfied regardless of where the secret technique is used.

2. **Extra Satisfied Customers via Secret Technique:**  
   The secret technique turns a window of size `minutes` from grumpy to non-grumpy. To maximize total satisfied customers, we need to find the contiguous window of length `minutes` that yields the maximum **additional** satisfied customers (i.e., customers entering when `grumpy[i] == 1`).

3. **Fixed Sliding Window:**  
   Use a sliding window of size `minutes` across the array to compute the gain in satisfied customers, keeping track of the `max_extra` customers recovered.

4. **Total Result:**  
   $\text{Total Satisfied} = \text{Baseline Satisfied} + \text{Maximum Extra Satisfied}$.


```python
class Solution:
    def maxSatisfied(self, customers, grumpy, minutes):

        # Customers already satisfied
        happy = 0

        for i in range(len(customers)):
            if grumpy[i] == 0:
                happy += customers[i]

        # Customers we can make happy in the first window
        extra = 0

        for i in range(minutes):
            if grumpy[i] == 1:
                extra += customers[i]

        max_extra = extra

        # Slide the window
        for i in range(minutes, len(customers)):

            # Add new customer
            if grumpy[i] == 1:
                extra += customers[i]

            # Remove old customer
            if grumpy[i - minutes] == 1:
                extra -= customers[i - minutes]

            max_extra = max(max_extra, extra)

        return happy + max_extra
```

```python
customers = [1,0,1,2,1,1,7,5]
grumpy = [0,1,0,1,0,1,0,1]
minutes = 3
result = Solution().maxSatisfied(customers, grumpy, minutes)
print(result)  # Output: 16
```

```python
customers = [1]
grumpy = [0]
minutes = 1
result = Solution().maxSatisfied(customers, grumpy, minutes)
print(result)  # Output: 1
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Single pass to calculate baseline and perform the sliding window.
* **Space Complexity:** $\mathcal{O}(1)$ — Constant extra space used.

```python
def egg_drop(floors):
    # Find the minimum number of drops needed
    drops = 0
    total = 0

    while total < floors:
        drops += 1
        total += drops

    print("Minimum worst-case drops:", drops)

    # Generate the floors for the first egg
    current_floor = 0
    step = drops

    while current_floor < floors:
        current_floor += step

        if current_floor > floors:
            current_floor = floors

        print("Drop first egg from floor:", current_floor)

        step -= 1

        if step == 0:
            break
```

```python
floors = int(input("Enter number of floors: "))

egg_drop(floors)
```

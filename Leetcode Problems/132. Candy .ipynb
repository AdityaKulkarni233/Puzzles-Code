# 135. Candy

## Problem Statement

There are $n$ children standing in a line. Each child is assigned a rating value given in the integer array `ratings`.

You are giving candies to these children subjected to the following requirements:

* Each child must have at least one candy.
* Children with a higher rating get more candies than their neighbors.

Return the *minimum number of candies* you need to have to distribute the candies to the children.

---

### Examples

#### Example 1
* **Input:** `ratings = [1,0,2]`
* **Output:** `5`
* **Explanation:** You can allocate to the first, second, and third child with `2, 1, 2` candies respectively.

#### Example 2
* **Input:** `ratings = [1,2,2]`
* **Output:** `4`
* **Explanation:** You can allocate to the first, second, and third child with `1, 2, 1` candies respectively.  
  *The third child gets 1 candy because it satisfies the above two conditions.*

---

### Constraints
* $n == \text{ratings.length}$
* $1 \le n \le 5 \times 10^4$
* $0 \le \text{ratings}[i] \le 5 \times 10^4$

---

## Optimal Solution (Two-Pass Greedy)

### Intuition & Approach

To satisfy the neighbor condition, we can split the requirement into two independent sub-problems:
1. **Left-to-Right Pass:** Ensure every child with a higher rating than their *left neighbor* gets more candies than that neighbor.
2. **Right-to-Left Pass:** Ensure every child with a higher rating than their *right neighbor* gets more candies than that neighbor.

#### Steps:
1. Initialize an array `candies` of length $n$ filled with `1` (since every child gets at least one candy).
2. **Left Pass:** Loop from left to right ($i = 1 \dots n-1$). If $\text{ratings}[i] > \text{ratings}[i-1]$, set $\text{candies}[i] = \text{candies}[i-1] + 1$.
3. **Right Pass:** Loop from right to left ($i = n-2 \dots 0$). If $\text{ratings}[i] > \text{ratings}[i+1]$, update $\text{candies}[i] = \max(\text{candies}[i], \text{candies}[i+1] + 1)$.
4. Return the sum of the `candies` array.

```python
def candy(ratings):
    n = len(ratings)
    if n == 0:
        return 0

    candies = [1] * n

    # Traverse from left to right
    for i in range(1, n):
        if ratings[i] > ratings[i - 1]:
            candies[i] = candies[i - 1] + 1

    # Traverse from right to left
    for i in range(n - 2, -1, -1):
        if ratings[i] > ratings[i + 1]:
            candies[i] = max(candies[i], candies[i + 1] + 1)

    return sum(candies)
```

```python
res = candy([1, 0, 2])
print(res)  # Output: 5
```

```python
res = candy([1, 2, 2])
print(res)  # Output: 4
```

```python
res = candy([1, 2, 4, 2, 1])
print(res)  # Output: 9
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Two passes over the array of size $n$.
* **Space Complexity:** $\mathcal{O}(n)$ — Space used for the `candies` allocation array.

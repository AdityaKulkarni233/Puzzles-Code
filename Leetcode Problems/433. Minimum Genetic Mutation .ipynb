# 433. Minimum Genetic Mutation

## Problem Statement

A gene string can be represented by an 8-character long string, with choices from `'A'`, `'C'`, `'G'`, and `'T'`.

Suppose we need to investigate a mutation from a gene string `startGene` to a gene string `endGene` where one mutation is defined as one single character changed in the gene string.

For example, `"AACCGGTT" --> "AACCGGTA"` is one mutation.

There is also a gene bank `bank` that records all the valid gene mutations. A gene must be in `bank` to make it a valid gene string.

Given the two gene strings `startGene` and `endGene` and the gene bank `bank`, return the *minimum number of mutations* needed to mutate from `startGene` to `endGene`. If there is no such a mutation, return `-1`.

**Note:** The starting point is assumed to be valid, so it might not be included in the bank.

---

### Examples

#### Example 1
* **Input:** `startGene = "AACCGGTT"`, `endGene = "AACCGGTA"`, `bank = ["AACCGGTA"]`
* **Output:** `1`

#### Example 2
* **Input:** `startGene = "AACCGGTT"`, `endGene = "AAACGGTA"`, `bank = ["AACCGGTA","AACCGCTA","AAACGGTA"]`
* **Output:** `2`

---

### Constraints
* $0 \le \text{bank.length} \le 10$
* $\text{startGene.length} == \text{endGene.length} == \text{bank}[i].length == 8$
* `startGene`, `endGene`, and `bank[i]` consist of only the characters `['A', 'C', 'G', 'T']`.

---

## Optimal Solution (Breadth-First Search / BFS)

### Intuition & Approach
Since we want to find the **minimum number of mutations** (shortest path) from `startGene` to `endGene`, **Breadth-First Search (BFS)** is the ideal algorithm.

1. **Graph Representation:**  
   Think of each 8-character gene string as a graph node. Two nodes share an edge if they differ by exactly **1 character** and the target string exists in `bank`.

2. **BFS Queue:**  
   - Store pairs of `(current_gene, mutation_steps)` in a double-ended queue (`deque`).
   - Keep a `visited` set to avoid revisiting genes and preventing infinite loops.

3. **Level-by-Level Traversal:**  
   - Pop the current gene from the queue.
   - If `current_gene == endGene`, return the current mutation count.
   - Try changing each of the 8 character positions to `'A'`, `'C'`, `'G'`, or `'T'`.
   - If the new string is in `bank` and has not been visited, mark it as visited and add it to the queue with `steps + 1`.

4. If the queue becomes empty and `endGene` was never reached, return `-1`.

```python

from collections import deque

def minMutation(startGene, endGene, bank):
    bank = set(bank)
    queue = deque([(startGene, 0)])
    genes = "ACGT"

    while queue:
        gene, steps = queue.popleft()

        if gene == endGene:
            return steps

        for i in range(8):
            for ch in genes:
                new_gene = gene[:i] + ch + gene[i+1:]

                if new_gene in bank:
                    bank.remove(new_gene)
                    queue.append((new_gene, steps + 1))

    return -1
```

```python
result = minMutation("AACCGGTT", "AAACGGTA", ["AACCGGTA", "AACCGCTA", "AAACGGTA"])
print(result)  # Output: 2
```

```python
result = minMutation("AACCGGTT", "AACCGGTA", ["AACCGGTA"])
print(result)  # Output: 1
```

### Complexity
* **Time Complexity:** $\mathcal{O}(N \cdot L \cdot \vert{}\Sigma\vert{})$  
  Where $N$ is the number of genes in the bank ($\le 10$), $L = 8$ is the length of each gene string, and $\vert{}\Sigma\vert{} = 4$ is the character set size (`'A'`, `'C'`, `'G'`, `'T'`). Since $N, L$ are tiny, this runs in $\mathcal{O}(1)$ time.
* **Space Complexity:** $\mathcal{O}(N)$ — Space to hold the bank set, visited set, and queue.

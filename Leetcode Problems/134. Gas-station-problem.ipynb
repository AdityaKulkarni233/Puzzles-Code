# ⛽ Gas Station Circuit Problem

Imagine you are driving a car around a circular road with several gas stations.

Each gas station gives you a certain amount of gas, but you also need some amount of gas to travel to the next station.

You start with an empty tank at one of the stations.

Your goal is to find the station from which you can start, collect gas along the way, and travel around the entire circular route exactly once, returning to the station where you started.

Your gas tank has unlimited capacity, so you can store as much gas as you want.

If it is impossible to complete the circuit from any station, return `-1`.

## Example 1

**Input:**

`gas = [1, 2, 3, 4, 5]`  
`cost = [3, 4, 5, 1, 2]`

**Output:**

`3`

**Explanation:**

Start at station `3` (index `3`) and fill up with `4` units of gas.

- Tank = `0 + 4 = 4`
- Travel to station `4` → Tank = `4 - 1 + 5 = 8`
- Travel to station `0` → Tank = `8 - 2 + 1 = 7`
- Travel to station `1` → Tank = `7 - 3 + 2 = 6`
- Travel to station `2` → Tank = `6 - 4 + 3 = 5`
- Travel back to station `3` → Cost = `5`, so the tank becomes `0`

The car successfully completes the entire circuit.

Therefore, return `3` as the starting index.

---

## Example 2

**Input:**

`gas = [2, 3, 4]`  
`cost = [3, 4, 3]`

**Output:**

`-1`

**Explanation:**

You cannot start at station `0` or `1` because there is not enough gas to travel to the next station.

Let's start at station `2` and fill up with `4` units of gas.

- Tank = `0 + 4 = 4`
- Travel to station `0` → Tank = `4 - 3 + 2 = 3`
- Travel to station `1` → Tank = `3 - 3 + 3 = 3`
- Travel back to station `2` requires `4` units of gas, but only `3` units remain.

Therefore, you cannot complete the circuit from any starting station.

**Return `-1`.**

<h4> Code solution and time complexity

<img src="https://i.ytimg.com/vi/9h-2YiTNam4/maxresdefault.jpg" width="1000" height="500" />

```python
class Solution:
    def canCompleteCircuit(self, gas: list[int], cost: list[int]) -> int:
        if sum(gas) < sum(cost):
            return -1
        tank , index = 0, 0
        for i in range(len(gas)):
            tank += gas[i]-cost[i]
            if tank < 0:
                tank ,index = 0 ,i+1
        return index
```

#### Example 1

```python
gas = [1, 2, 3, 4, 5]
cost = [3, 4, 5, 1, 2]

solution = Solution()
result = solution.canCompleteCircuit(gas, cost)
print(result)
```

#### Example 2

```python
gas = [2, 3, 4]
cost = [3, 4, 3]
solution = Solution()
result = solution.canCompleteCircuit(gas, cost)
print(result)
```

## Complexity

### Time Complexity

**O(n)**

- We loop through the `gas` array once.
- `sum(gas)` and `sum(cost)` each take O(n).
- Overall time complexity is **O(n)**.

### Space Complexity

**O(1)**

- We only use a few variables: `tank`, `index`, and `i`.
- No extra data structure is created.
- Therefore, space complexity is **O(1)**.

# Best Time to Buy and Sell Stock II

## Problem Statement

You are given an integer array `prices` where `prices[i]` is the price of a given stock on the $i^{\text{th}}$ day.

On each day, you may decide to buy and/or sell the stock. You can only hold **at most one share** of the stock at any time. However, you can buy it and then immediately sell it on the same day.

Find and return the *maximum profit* you can achieve.

---

### Examples

#### Example 1
* **Input:** `prices = [7,1,5,3,6,4]`
* **Output:** `7`
* **Explanation:** Buy on day 2 (price = 1) and sell on day 3 (price = 5), profit = $5 - 1 = 4$.  
  Then buy on day 4 (price = 3) and sell on day 5 (price = 6), profit = $6 - 3 = 3$.  
  Total profit is $4 + 3 = 7$.

#### Example 2
* **Input:** `prices = [1,2,3,4,5]`
* **Output:** `4`
* **Explanation:** Buy on day 1 (price = 1) and sell on day 5 (price = 5), profit = $5 - 1 = 4$.  
  Total profit is $4$.

#### Example 3
* **Input:** `prices = [7,6,4,3,1]`
* **Output:** `0`
* **Explanation:** There is no way to make a positive profit, so we never buy the stock to achieve the maximum profit of `0`.

---

### Constraints
* $1 \le \text{prices.length} \le 3 \times 10^4$
* $0 \le \text{prices}[i] \le 10^4$

---

## Optimal Solution (Greedy / One-Pass)

### Intuition & Approach
Since you can complete as many transactions as you like, you want to capture **every single positive price increase** from one day to the next.

For any consecutive pair of days where $\text{prices}[i] > \text{prices}[i - 1]$, you add the difference $\text{prices}[i] - \text{prices}[i - 1]$ to your total profit.

This holds true because holding a stock across multiple upward days is mathematically identical to buying and selling every day during the rise:
$$\text{prices}[3] - \text{prices}[1] = (\text{prices}[2] - \text{prices}[1]) + (\text{prices}[3] - \text{prices}[2])$$

```python
def maxProfit(prices: list[int]) -> int:
    max_profit = 0
    
    for i in range(1, len(prices)):
        if prices[i] > prices[i - 1]:
            max_profit += prices[i] - prices[i - 1]
            
    return max_profit
```

```python
mp = maxProfit([7, 1, 5, 3, 6, 4])
print(mp)
```

```python
mp = maxProfit([1,2,3,4,5])
print(mp)
```

```python
mp=maxProfit([7,6,4,3,1])
print(mp)
```


### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — We iterate through the array once.
* **Space Complexity:** $\mathcal{O}(1)$ — Constant extra space used.

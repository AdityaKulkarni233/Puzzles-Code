#### Import and constants

```python
import random

RED = "Red"
BLUE = "Blue"
```

#### Create 100 random hats

```python
hats = []

for i in range(100):
    hats.append(random.choice([RED, BLUE]))

print(hats)
```

#### Create the solution function

```python
def solve(hats):

    n = len(hats)
    answers = []

    # Prisoner 1 counts the Red hats in front
    red_count = 0

    for i in range(1, n):
        if hats[i] == RED:
            red_count += 1

    # Red means even
    # Blue means odd
    if red_count % 2 == 0:
        answers.append(RED)
        parity = 0
    else:
        answers.append(BLUE)
        parity = 1

    # Remaining prisoners
    for i in range(1, n):

        # Count Red hats ahead
        red_ahead = 0

        for j in range(i + 1, n):
            if hats[j] == RED:
                red_ahead += 1

        # Count Red answers already given
        red_answers = 0

        for answer in answers:
            if answer == RED:
                red_answers += 1

        # Determine own hat
        if (red_ahead + red_answers) % 2 == parity:
            answers.append(BLUE)
        else:
            answers.append(RED)

    return answers
```

#### Get the prisoners' answers

```python
answers = solve(hats)
print(answers)
```

#### Check who survives

```python
survivors = 0

for i in range(100):

    if hats[i] == answers[i]:
        status = "Survives"
        survivors += 1
    else:
        status = "Dies"

    print(
        "Prisoner", i + 1,
        "| Hat:", hats[i],
        "| Says:", answers[i],
        "|", status
    )
```

#### Final result

```python
print("Survivors:", survivors, "/ 100")
```

The first prisoner can die, because their answer is being used to communicate the parity information. </br>
The remaining 99 are guaranteed to get their own hat correct. </br>
But since the condition is if all the 99 prisoners guess correctly and if he(1st prisoner the last) answers wrong even tough he will be free to go with all of the remaining prisoners.

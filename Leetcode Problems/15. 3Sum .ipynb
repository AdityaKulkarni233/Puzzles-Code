# 15. 3Sum

## Problem Statement

Given an integer array `nums`, return all the triplets `[nums[i], nums[j], nums[k]]` such that $i \neq j$, $i \neq k$, and $j \neq k$, and `nums[i] + nums[j] + nums[k] == 0`.

Notice that the solution set **must not contain duplicate triplets**.

---

### Examples

#### Example 1
* **Input:** `nums = [-1,0,1,2,-1,-4]`
* **Output:** `[[-1,-1,2],[-1,0,1]]`
* **Explanation:**  
  `nums[0] + nums[1] + nums[2] = (-1) + 0 + 1 = 0`  
  `nums[1] + nums[2] + nums[4] = 0 + 1 + (-1) = 0`  
  `nums[0] + nums[3] + nums[4] = (-1) + 2 + (-1) = 0`  
  The distinct triplets are `[-1,-1,2]` and `[-1,0,1]`.

#### Example 2
* **Input:** `nums = [0,1,1]`
* **Output:** `[]`
* **Explanation:** The only possible triplet does not sum up to 0.

#### Example 3
* **Input:** `nums = [0,0,0]`
* **Output:** `[[0,0,0]]`
* **Explanation:** The only possible triplet sums up to 0.

---

### Constraints
* $3 \le \text{nums.length} \le 3000$
* $-10^5 \le \text{nums}[i] \le 10^5$

---

## Brute Force Approach

### Intuition & Strategy
Use three nested loops to test every unique combination of three indices `(i, j, k)`. To prevent duplicate triplets in the output, sort each valid triplet and add it to a set before returning the results as a list.

```python
class Solution:
    def threeSum(self, nums: list[int]) -> list[list[int]]:
        n = len(nums)
        res = set()
        
        for i in range(n - 2):
            for j in range(i + 1, n - 1):
                for k in range(j + 1, n):
                    if nums[i] + nums[j] + nums[k] == 0:
                        triplet = tuple(sorted([nums[i], nums[j], nums[k]]))
                        res.add(triplet)
                        
        return [list(t) for t in res]
```

```python
sol = Solution()
res = sol.threeSum([-1,0,1,2,-1,-4])
print(res)
```

```python
sol = Solution()
res = sol.threeSum([0,1,1])
print(res)
```

```python
sol = Solution()
res = sol.threeSum([0,0,0])
print(res)
```

### Complexity Analysis
* **Time Complexity:** $\mathcal{O}(n^3)$ — Three nested loops scanning all combinations.
* **Space Complexity:** $\mathcal{O}(n)$ — Space used by the set to store unique triplets.

# 502. IPO

## Problem Statement

Suppose LeetCode will start its IPO soon. In order to sell a good price of its shares to Venture Capital, LeetCode would like to work on some projects to increase its capital before the IPO. Since it has limited resources, it can only finish **at most $k$ distinct projects** before the IPO. Help LeetCode design the best way to maximize its total capital after finishing at most $k$ distinct projects.

You are given $n$ projects where the $i^{\text{th}}$ project has a pure profit `profits[i]` and a minimum capital `capital[i]` needed to start it.

Initially, you have $w$ capital. When you finish a project, you will obtain its pure profit and the profit will be added to your total capital.

Pick a list of **at most $k$ distinct projects** from given projects to maximize your final capital, and return the final maximized capital.

The answer is guaranteed to fit in a 32-bit signed integer.

---

### Examples

#### Example 1
* **Input:** `k = 2`, `w = 0`, `profits = [1,2,3]`, `capital = [0,1,1]`
* **Output:** `4`
* **Explanation:**  
  1. Initial capital is `0`, so you can only start project index `0`. Your capital becomes `0 + 1 = 1`.
  2. With capital `1`, you can choose project index `1` or project index `2`. To maximize profit, pick project `2`.
  3. Final capital becomes `1 + 3 = 4`.

#### Example 2
* **Input:** `k = 3`, `w = 0`, `profits = [1,2,3]`, `capital = [0,1,2]`
* **Output:** `6`

---

### Constraints
* $1 \le k \le 10^5$
* $0 \le w \le 10^9$
* $n == \text{profits.length} == \text{capital.length}$
* $1 \le n \le 10^5$
* $0 \le \text{profits}[i] \le 10^4$
* $0 \le \text{capital}[i] \le 10^9$

---

## Optimal Solution (Greedy + Min-Heap & Max-Heap)

### Intuition & Approach
To maximize capital at each step, we should always pick the available project with the **highest profit** that we can currently afford (`capital[i] <= w`).

1. **Sort Projects:** Pair `(capital[i], profits[i])` together and sort projects by their required capital in ascending order.
2. **Max-Heap for Affordable Profits:** Maintain a Max-Heap to store profits of all projects we can afford with our current capital $w$.
3. **Greedy Selection:**
   - Iterate at most $k$ times.
   - Push profits of all affordable projects (`capital <= w`) into the Max-Heap.
   - If the Max-Heap is empty, stop early (no more projects can be started).
   - Pop the top (maximum) profit from the Max-Heap and add it to current capital $w$.

```python
import heapq
class Solution:
     def findMaximizedCapital(self, k, w, profits, capital):
        n = len(profits)
        projects = sorted(zip(capital, profits)) # Pair capital and profits, then sort by capital ascending
        
        max_profit_heap = []
        i = 0
        for _ in range(k):
            # Add all projects we can currently afford into the max-heap
            while i < n and projects[i][0] <= w:
                heapq.heappush(max_profit_heap, -projects[i][1])  # Min-heap simulated as max-heap
                i += 1
            
            # If no affordable projects are available, return current capital
            if not max_profit_heap:
                break     
            # Pop the project with the highest profit
            w += -heapq.heappop(max_profit_heap)      
        return w
```

```python
solution = Solution()
# Example usage:

k = 2
w = 0
profits = [1, 2, 3]
capital = [0, 1, 1]
result = solution.findMaximizedCapital(k, w, profits, capital)
print(result)  # Output: 4
```

```python
solution = Solution()
# Example 2 usage:
k = 3
w = 0
profits = [1, 2, 3]
capital = [0, 1, 2]
result = solution.findMaximizedCapital(k, w, profits, capital)
print(result)  # Output: 6
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n \log n + k \log n)$ — Sorting takes $\mathcal{O}(n \log n)$ and each heap operation takes $\mathcal{O}(\log n)$.
* **Space Complexity:** $\mathcal{O}(n)$ — To store projects and heap elements.

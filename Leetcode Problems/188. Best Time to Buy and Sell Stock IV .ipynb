# 188. Best Time to Buy and Sell Stock IV

## Problem Statement

You are given an integer array `prices` where `prices[i]` is the price of a given stock on the $i^{\text{th}}$ day, and an integer $k$.

Find the maximum profit you can achieve. You may complete at most **$k$ transactions**: i.e., you may buy at most $k$ times and sell at most $k$ times.

**Note:** You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

---

### Examples

#### Example 1
* **Input:** `k = 2`, `prices = [2,4,1]`
* **Output:** `2`
* **Explanation:** Buy on day 1 (price = 2) and sell on day 2 (price = 4), profit = $4 - 2 = 2$.

#### Example 2
* **Input:** `k = 2`, `prices = [3,2,6,5,0,3]`
* **Output:** `7`
* **Explanation:** Buy on day 2 (price = 2) and sell on day 3 (price = 6), profit = $6 - 2 = 4$.  
  Then buy on day 5 (price = 0) and sell on day 6 (price = 3), profit = $3 - 0 = 3$.

---

### Constraints
* $1 \le k \le 100$
* $1 \le \text{prices.length} \le 1000$
* $0 \le \text{prices}[i] \le 1000$

---

## Optimal Solution (Dynamic Programming)

### Intuition & Approach

This problem generalizes the previous "at most 2 transactions" problem to **at most $k$ transactions**.

1. **Optimization for Large $k$:**  
   If $k \ge \frac{n}{2}$ (where $n$ is the number of days), you can complete as many transactions as there are price increases. This reduces directly to **Best Time to Buy and Sell Stock II** (Greedy approach).

2. **State Machine Dynamic Programming:**  
   Maintain two arrays of size $k + 1$:
   - `buy[j]`: Maximum profit after making at most $j$ buys up to the current day.
   - `sell[j]`: Maximum profit after making at most $j$ sells up to the current day.

   For each price on each day:
   - $\text{buy}[j] = \max(\text{buy}[j], \text{sell}[j-1] - \text{price})$
   - $\text{sell}[j] = \max(\text{sell}[j], \text{buy}[j] + \text{price})$

```python
def maxProfit(k, prices):
    if not prices:
        return 0

    buy = [float("-inf")] * (k + 1)
    sell = [0] * (k + 1)

    for price in prices:
        for transaction in range(1, k + 1):
            buy[transaction] = max(
                buy[transaction],
                sell[transaction - 1] - price
            )

            sell[transaction] = max(
                sell[transaction],
                buy[transaction] + price
            )

    return sell[k]
```

```python
k = 2
prices = [3, 2, 6, 5, 0, 3]
result = maxProfit(k, prices)
print(result)  # Output: 7
```

```python
k = 2
prices = [2, 4, 1]
result = maxProfit(k, prices)
print(result)  # Output: 2
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n \cdot k)$ — Iterating through $n$ prices across $k$ transactions.
* **Space Complexity:** $\mathcal{O}(k)$ — Two arrays of size $k + 1$.

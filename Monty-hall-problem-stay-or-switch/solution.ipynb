```python
import random

# Put the car behind one of the 3 doors
car = random.randint(1, 3)
```

```python
print("Monty Hall Problem")
print("1. Door 1")
print("2. Door 2")
print("3. Door 3")
```

```python
choice = int(input("Choose a door: "))
```

```python
# Monty opens a door with a goat
if choice != car:
    opened = 6 - choice - car
else:
    goats = [door for door in [1, 2, 3] if door != choice]
    opened = random.choice(goats)

print(f"Monty opens Door {opened} then it is  🐐")
```

```python
# The remaining unopened door
switch = 6 - choice - opened

answer = input(f"Stay with Door {choice} or switch to Door {switch}? ").lower()

if answer == "switch":
    choice = switch
```

```python
# Reveal the result
if choice == car:
    print("🚗 You won the car!")
else:
    print("🐐 You got a goat!")

print(f"The car was behind Door {car}.")
```

Expected Probability

In a single game:

Stay: 1/3 ≈ 33.33% chance of winning 🚗 </br>
Switch: 2/3 ≈ 66.67% chance of winning 🚗

You can test the program by playing multiple times and recording your results. Over many independent games, the observed win rates should generally move toward these theoretical probabilities.

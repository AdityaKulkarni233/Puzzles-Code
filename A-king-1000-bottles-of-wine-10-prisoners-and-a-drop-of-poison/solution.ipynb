```python
import random

bottles  = 1000
prisoners  = 10  #2^10 = 1024
```

```python
def test_bottle(bottle):
    dead = []
    for prisoner in range(prisoners): # Check whether this prisoner's bit is 1.
        if bottle & (1 << prisoner):
            dead.append(prisoner)
    return dead
```

```python
def find_bottle(dead):
    bottle = 0
    for prisoner in dead:
        bottle += 2**prisoner # Add the value represented by this prisoner.
    return bottle
```

```python
# Test every bottle
for bottle in range(bottles): # Find which prisoners drink from this bottle.
    dead = test_bottle(bottle) # Use those prisoners to identify the bottle.
    result = find_bottle(dead) # The result should always match the original bottle.

    assert result == bottle
```

```python
poisoned = random.randint(0, 999) # Example

dead = test_bottle(poisoned) # Find which prisoners drink from the poisoned bottle.
result = find_bottle(dead) # Identify the poisoned bottle from the dead prisoners.

print("Poisoned bottle:", poisoned)
print("Dead prisoners:", dead)
print("Found bottle:", result)
```

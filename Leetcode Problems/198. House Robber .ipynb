# 198. House Robber

## Problem Statement

You are a professional robber planning to rob houses along a street. Each house has a certain amount of money stashed. The only constraint stopping you from robbing each of them is that adjacent houses have security systems connected and **it will automatically contact the police if two adjacent houses were broken into on the same night**.

Given an integer array `nums` representing the amount of money of each house, return the *maximum amount of money you can rob tonight without alerting the police*.

---

### Examples

#### Example 1
* **Input:** `nums = [1,2,3,1]`
* **Output:** `4`
* **Explanation:** Rob house 1 (money = 1) and then rob house 3 (money = 3).  
  Total amount you can rob = $1 + 3 = 4$.

#### Example 2
* **Input:** `nums = [2,7,9,3,1]`
* **Output:** `12`
* **Explanation:** Rob house 1 (money = 2), rob house 3 (money = 9) and rob house 5 (money = 1).  
  Total amount you can rob = $2 + 9 + 1 = 12$.

---

### Constraints
* $1 \le \text{nums.length} \le 100$
* $0 \le \text{nums}[i] \le 400$

---

## Optimal Solution (Dynamic Programming with $\mathcal{O}(1)$ Space)

### Intuition & Approach
For each house $i$, you have two choices:
1. **Rob house $i$:** You gain `nums[i]`, but you cannot rob house $i-1$. So your total becomes `nums[i] + rob_prev2`.
2. **Skip house $i$:** Your total remains equal to the maximum stolen up to house $i-1$ (`rob_prev1`).

The decision recurrence is:
$$\text{dp}[i] = \max(\text{dp}[i-1], \text{dp}[i-2] + \text{nums}[i])$$

Since we only ever need the maximum values from the **last two houses** (`rob_prev1` and `rob_prev2`), we don't need a full array. We can keep track of them using two variables in $\mathcal{O}(1)$ space.

```python
class Solution:
    def rob(self, nums: list[int]) -> int:
        rob_prev1 = 0  # Max profit up to house i-1
        rob_prev2 = 0  # Max profit up to house i-2
        
        for num in nums:
            current = max(rob_prev1, rob_prev2 + num)
            rob_prev2 = rob_prev1
            rob_prev1 = current
            
        return rob_prev1
```

```python
sol = Solution()
res = sol.rob([1,2,3,1])
print(res)
```

```python
sol = Solution()
res = sol.rob([2,7,9,3,1])
print(res)
```


### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Single loop through the `nums` array.
* **Space Complexity:** $\mathcal{O}(1)$ — Constant extra space used.

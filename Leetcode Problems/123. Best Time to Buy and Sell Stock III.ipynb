# 123. Best Time to Buy and Sell Stock III

## Problem Statement

You are given an array `prices` where `prices[i]` is the price of a given stock on the $i^{\text{th}}$ day.

Find the maximum profit you can achieve. You may complete **at most two transactions**.

**Note:** You may not engage in multiple transactions simultaneously (i.e., you must sell the stock before you buy again).

---

### Examples

#### Example 1
* **Input:** `prices = [3,3,5,0,0,3,1,4]`
* **Output:** `6`
* **Explanation:** Buy on day 4 (price = 0) and sell on day 6 (price = 3), profit = $3 - 0 = 3$.  
  Then buy on day 7 (price = 1) and sell on day 8 (price = 4), profit = $4 - 1 = 3$.

#### Example 2
* **Input:** `prices = [1,2,3,4,5]`
* **Output:** `4`
* **Explanation:** Buy on day 1 (price = 1) and sell on day 5 (price = 5), profit = $5 - 1 = 4$.  
  *Note that you cannot buy on day 1, buy on day 2 and sell them later, as you are engaging in multiple transactions at the same time. You must sell before buying again.*

#### Example 3
* **Input:** `prices = [7,6,4,3,1]`
* **Output:** `0`
* **Explanation:** In this case, no transaction is done, i.e., max profit = `0`.

---

### Constraints
* $1 \le \text{prices.length} \le 10^5$
* $0 \le \text{prices}[i] \le 10^5$

---

## Optimal Solution (Dynamic Programming / State Machine)

### Intuition & Approach
We can model this problem using four variables representing the maximum net profit (or cash on hand) after each state on any given day:

1. `first_buy`: Profit after buying the 1st stock (we minimize money spent, so it represents effective negative cost).
2. `first_sell`: Profit after selling the 1st stock.
3. `second_buy`: Profit after buying the 2nd stock (using profit gained from the 1st transaction).
4. `second_sell`: Profit after selling the 2nd stock.

For each price in `prices`:
- $\text{first\_buy} = \max(\text{first\_buy}, -\text{price})$
- $\text{first\_sell} = \max(\text{first\_sell}, \text{first\_buy} + \text{price})$
- $\text{second\_buy} = \max(\text{second\_buy}, \text{first\_sell} - \text{price})$
- $\text{second\_sell} = \max(\text{second\_sell}, \text{second\_buy} + \text{price})$

```python
def maxProfit(prices: list[int]) -> int:
    first_buy = float('-inf')
    first_sell = 0
    second_buy = float('-inf')
    second_sell = 0

    for price in prices:
        first_buy = max(first_buy, -price)
        first_sell = max(first_sell, first_buy + price)
        second_buy = max(second_buy, first_sell - price)
        second_sell = max(second_sell, second_buy + price)

    return second_sell
```

```python
mp = maxProfit([3,3,5,0,0,3,1,4])
print(mp)  # Output: 6
```

```python
mp = maxProfit([1,2,3,4,5])
print(mp)
```

```python
mp = maxProfit([7,6,4,3,1]) # here no transaction is done, i.e. max profit = 0
print(mp)
```

### Complexity
* **Time Complexity:** $\mathcal{O}(n)$ — Single pass through the array.
* **Space Complexity:** $\mathcal{O}(1)$ — Constant extra space used.
